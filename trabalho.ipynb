# relação entre gêneros de animes mais populares e os animes que mais influenciam a cultura japonesa
# Limpeza: a função limpar_genero tira o prefixo (Theme::, Demographic::) e detecta quando a string é literalmente repetida ("X X"), devolvendo só "X". Testei nos 76 gêneros únicos e todos ficaram limpos.
# Gráfico 1: junta cada anime aos seus gêneros, soma o members por gênero e mostra os 10 com maior audiência total. No seu dataset, deu Action, Shounen e Fantasy no topo.
# Gráfico 2: em vez de um bar chart (testei — como os top 10 ficam todos entre percentil 99,5 e 99,98, as barras saem quase idênticas e não dizem nada visualmente), usei um scatter: nota no eixo X, membros no eixo Y, com todos os animes em cinza de fundo e os 10 mais "influentes" em vermelho com o nome anotado. Assim dá pra ver que eles estão no canto superior direito (nota alta + audiência alta) — o motivo real por trás do ranking fica visível no gráfico, não escondido dentro de um número só.
# Analisando o anime_genres.csv descobri um problema de qualidade nos dados: o campo genre vem duplicado (ex: "Action Action", "Theme::Isekai Isekai") — parece um bug de raspagem do site. 

```python
import pandas as pd
import re
import matplotlib.pyplot as plt

# 1. Carregar os dados
anime = pd.read_csv('anime.csv')
generos = pd.read_csv('anime_genres.csv')
anime['ano_lancamento'] = pd.to_datetime(anime['start_date'], errors='coerce').dt.year.astype('Int64')

# 2. Limpar a coluna 'genre' (vem duplicada: "Action Action", "Theme::Isekai Isekai")
def limpar_genero(g):
    g = re.sub(r'^[A-Za-z]+::', '', g)          # remove prefixo tipo 'Theme::' / 'Demographic::'
    n = len(g)
    if n % 2 == 1:
        meio = n // 2
        primeira, segunda = g[:meio], g[meio + 1:]
        if primeira == segunda:                 # string repetida "X X"
            return primeira
    return g

generos['genre'] = generos['genre'].apply(limpar_genero)

# ===================== GRÁFICO 1: gêneros mais populares =====================
# "popularidade do gênero" = soma dos membros de todos os animes que têm esse gênero
anime_genero = generos.merge(anime[['anime_id', 'members']], on='anime_id')
genero_pop = (
    anime_genero.groupby('genre')['members']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
genero_pop.sort_values().plot(kind='barh', color='mediumseagreen')
plt.xlabel('Total de membros (soma de todos os animes do gênero)')
plt.ylabel('')
plt.title('Top 10 gêneros mais populares')
plt.tight_layout()
plt.show()

# ===================== GRÁFICO 2: proxy de "influência cultural" =====================
# Combina nota e audiência via percentil (0 a 1), pra não deixar 'members' dominar
anime['percentil_score'] = anime['score'].rank(pct=True)
anime['percentil_members'] = anime['members'].rank(pct=True)
anime['influencia_cultural'] = (anime['percentil_score'] + anime['percentil_members']) / 2

top_influencia = anime.sort_values('influencia_cultural', ascending=False).head(10)

plt.figure(figsize=(10, 7))
plt.scatter(anime['score'], anime['members'], alpha=0.15, color='gray', s=15, label='Todos os animes')
plt.scatter(top_influencia['score'], top_influencia['members'], color='indianred', s=70, label='Top 10 influência')
for _, row in top_influencia.iterrows():
    plt.annotate(row['title'], (row['score'], row['members']), fontsize=8, xytext=(6, 3), textcoords='offset points')
plt.xlabel('Nota média (score)')
plt.ylabel('Número de membros')
plt.title('Nota vs. audiência - top 10 animes com maior influência aproximada')
plt.legend()
plt.tight_layout()
plt.show()
```

# Quais estúdios de animação entregam a melhor nota média (entre os que já produziram um volume relevante)? — ajuda a identificar parceiros/estúdios mais consistentes pra investir ou licenciar conteúdo

# Pergunta 1: junta anime_companies (filtrando role == 'Studio') com entities (pra trazer o nome do estúdio) e com anime (pra pegar a nota). Agrupa por estúdio, exige no mínimo 5 produções (senão um estúdio com 1 anime nota 9 aparece artificialmente no topo) e pega os 10 melhores. No dataset, nomes como Studio Signpost, Sharefun Studio e Studio Bind lideram.




```python
import pandas as pd
import matplotlib.pyplot as plt

anime = pd.read_csv('anime.csv')
anime['ano_lancamento'] = pd.to_datetime(anime['start_date'], errors='coerce').dt.year.astype('Int64')

companies = pd.read_csv('anime_companies.csv')
entities = pd.read_csv('entities.csv')

# Filtra so o papel 'Studio' (produtora de animacao, diferente de 'Producer'/'Licensor')
estudios = companies[companies['role'] == 'Studio'].merge(
    entities[['entity_id', 'name']], left_on='company_id', right_on='entity_id'
)
estudios_anime = estudios.merge(anime[['anime_id', 'score']], on='anime_id')

resumo_estudio = (
    estudios_anime.groupby('name')
    .agg(qtd_animes=('anime_id', 'count'), nota_media=('score', 'mean'))
    .query('qtd_animes >= 5')          # remove estudios com poucas producoes (ruido estatistico)
    .sort_values('nota_media', ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
resumo_estudio.sort_values('nota_media')['nota_media'].plot(kind='barh', color='teal')
plt.xlim(7, 8.3)  # zoom no intervalo real dos dados, pra diferenca ficar visivel
plt.xlabel('Nota media dos animes produzidos')
plt.ylabel('')
plt.title('Top 10 estudios com melhor nota media (min. 5 producoes)')
plt.tight_layout()
plt.show()
```

# Qual formato (TV, filme, OVA, ONA, especial) traz mais audiência e melhor nota, em média? — orienta em que tipo de produção vale mais apostar. 

# Pergunta 2: agrupa direto no anime.csv por type. TV lidera tanto em audiência média quanto em nota média — ou seja, séries de TV não são só mais vistas, também são um pouco melhor avaliadas que filme/OVA/especial, em média.

# O volume de lançamentos por ano está crescendo, e a qualidade média está acompanhando esse crescimento? — indica se o mercado está saturando (mais oferta, qualidade caindo) ou se está aquecido e saudável

# Pergunta 3: usa ax2 = ax1.twinx() pra colocar duas escalas diferentes (quantidade e nota) no mesmo gráfico, com eixo X compartilhado — uma técnica útil pra comparar volume e qualidade lado a lado. Nos  dados, o volume de lançamentos cresce praticamente todo ano desde os anos 1990, e a nota média também sobe de leve (6.4 pra 7.1) — sinal de que o mercado não está "diluindo" qualidade ao crescer.

```python
import pandas as pd
import matplotlib.pyplot as plt

anime = pd.read_csv('anime.csv')
anime['ano_lancamento'] = pd.to_datetime(anime['start_date'], errors='coerce').dt.year.astype('Int64')

anime_por_ano = anime.dropna(subset=['ano_lancamento']).groupby('ano_lancamento').agg(
    qtd_lancamentos=('anime_id', 'count'),
    nota_media=('score', 'mean')
)

fig, ax1 = plt.subplots(figsize=(12, 6))
ax1.bar(anime_por_ano.index.astype(int), anime_por_ano['qtd_lancamentos'], color='steelblue', alpha=0.6)
ax1.set_xlabel('Ano de lançamento')
ax1.set_ylabel('Quantidade de lançamentos', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

ax2 = ax1.twinx()  # segundo eixo Y, compartilhando o mesmo eixo X
ax2.plot(anime_por_ano.index.astype(int), anime_por_ano['nota_media'], color='darkred', linewidth=2)
ax2.set_ylabel('Nota média', color='darkred')
ax2.tick_params(axis='y', labelcolor='darkred')

plt.title('Volume de lançamentos vs. nota média por ano')
fig.tight_layout()
plt.show()
```

# Matriz de correlação entre as variáveis numéricas — mostra quais colunas carregam informação redundante antes de você escolher features.
# (correlação): score e rank são quase opostos perfeitos (-0.98, esperado — rank é derivado da nota). episodes não se correlaciona com nada (0.03-0.04) — quantidade de episódios praticamente não prediz nota nem popularidade nesse dataset, algo bom de saber antes de usar episodes como feature.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

anime = pd.read_csv('anime.csv')

cols = ['score', 'rank', 'popularity', 'members', 'episodes']
corr = anime[cols].corr()

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(cols))); ax.set_yticks(range(len(cols)))
ax.set_xticklabels(cols, rotation=45, ha='right')
ax.set_yticklabels(cols)
for i in range(len(cols)):
    for j in range(len(cols)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha='center', va='center')
plt.colorbar(im, label='Correlação')
plt.title('Correlação entre variáveis numéricas')
plt.tight_layout()
plt.show()
```

# Distribuição de nota e de membros — entender a forma dos dados (assimetria) ajuda a decidir se precisa de transformação (ex: log) antes de modelar.

# Diretores com melhor nota média (mín. 5 produções) — quem entrega consistência de qualidade
# (diretores): Miyawaki Chizuru lidera com nota média 8,58 em 10 produções — bem acima do segundo colocado. Ikuhara Kunihiko também aparece no top 10, o que bate com a fama dele no meio (diretor de Revolutionary Girl Utena, Sarazanmai etc.) — bom sinal de que os dados fazem sentido

```python
import pandas as pd
import re
import matplotlib.pyplot as plt

anime = pd.read_csv('anime.csv')

staff = pd.read_csv('anime_staff.csv')
entities = pd.read_csv('entities.csv')

# pega so quem tem 'Director' como papel principal (exclui 'Episode Director', que e outro cargo)
diretores = staff[(staff['role'] == 'Director') | staff['role'].str.startswith('Director,', na=False)]
diretores = diretores.merge(entities[['entity_id', 'name']], left_on='person_id', right_on='entity_id')
diretores_anime = diretores.merge(anime[['anime_id', 'score']], on='anime_id').drop_duplicates(subset=['anime_id', 'name'])

resumo_diretor = (
    diretores_anime.groupby('name')
    .agg(qtd_animes=('anime_id', 'count'), nota_media=('score', 'mean'))
    .query('qtd_animes >= 5')          # minimo de producoes pra evitar ruido estatistico
    .sort_values('nota_media', ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
resumo_diretor.sort_values('nota_media')['nota_media'].plot(kind='barh', color='darkcyan')
plt.xlim(7, 8.8)
plt.xlabel('Nota média dos animes dirigidos')
plt.ylabel('')
plt.title('Top 10 diretores com melhor nota média (mín. 5 produções)')
plt.tight_layout()
plt.show()
```

# Gêneros por qualidade (nota), não por popularidade — complementa o gráfico de gêneros que fizemos antes: o que é mais assistido nem sempre é o mais bem avaliado
# (gêneros por qualidade): o retrato muda bastante em relação ao gráfico de popularidade de antes. "Award Winning", "Iyashikei" (gênero de anime relaxante/reconfortante) e "Organized Crime" têm a melhor nota média — nenhum deles apareceu no top de popularidade. Já "Kids" e "Ecchi" ficam no fundo da tabela de qualidade. Ou seja: gênero mais assistido ≠ gênero mais bem avaliado, uma distinção importante se seu modelo for prever "sucesso" (você precisa decidir se sucesso = audiência ou = nota).

```python
import pandas as pd
import re
import matplotlib.pyplot as plt

anime = pd.read_csv('anime.csv')

generos = pd.read_csv('anime_genres.csv')

def limpar_genero(g):
    g = re.sub(r'^[A-Za-z]+::', '', g)
    n = len(g)
    if n % 2 == 1:
        meio = n // 2
        p1, p2 = g[:meio], g[meio + 1:]
        if p1 == p2:
            return p1
    return g

generos['genre'] = generos['genre'].apply(limpar_genero)
anime_genero = generos.merge(anime[['anime_id', 'score']], on='anime_id')

qualidade_genero = (
    anime_genero.groupby('genre')
    .agg(qtd=('anime_id', 'count'), nota_media=('score', 'mean'))
    .query('qtd >= 30')                # generos raros de fora, media pouco confiavel
    .sort_values('nota_media', ascending=False)
)

top8 = qualidade_genero.head(8)
bottom5 = qualidade_genero.tail(5)
combinado = pd.concat([bottom5, top8]).sort_values('nota_media')
cores = ['indianred' if nome in bottom5.index else 'seagreen' for nome in combinado.index]

plt.figure(figsize=(10, 7))
plt.barh(combinado.index, combinado['nota_media'], color=cores)
plt.xlim(6, 8)
plt.xlabel('Nota média')
plt.title('Gêneros: melhor nota média (verde) vs. pior nota média (vermelho)')
plt.tight_layout()
plt.show()
```
