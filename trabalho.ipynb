```python
import pandas as pd

# 1. Carregar o dataset principal
anime = pd.read_csv('anime.csv')  # ajuste o caminho se o arquivo estiver em outra pasta

# 2. Extrair o ano de lançamento a partir de start_date (formato AAAA-MM-DD)
anime['ano_lancamento'] = pd.to_datetime(anime['start_date'], errors='coerce').dt.year.astype('Int64')

# 3. Selecionar os N animes mais populares
# 'popularity' é um ranking do MAL: quanto MENOR o valor, mais popular (1 = mais popular)
top_n = 10
mais_populares = anime.sort_values('popularity').head(top_n)[
    ['title', 'ano_lancamento', 'popularity', 'members', 'score']
]

print(mais_populares.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# 4. Gráfico de barras horizontais
plt.figure(figsize=(10, 6))
plt.barh(mais_populares['title'], mais_populares['members'], color='steelblue')
plt.gca().invert_yaxis()  # deixa o #1 no topo
plt.xlabel('Número de membros (usuários que adicionaram o anime à lista)')
plt.title(f'Top {top_n} animes mais populares')
plt.tight_layout()
plt.show()
```
